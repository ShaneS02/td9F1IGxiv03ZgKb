```python
from models.CoreModel import CoreModel
```

```python
model = CoreModel()
model.load_data_from_csv("../data/ACME-HappinessSurvey2020.csv")
model.preprocess_data()

    # split the dataset into training and testing sets
X_train, X_test, y_train, y_test = model.split_data()
X_train_scaled, X_test_scaled = model.scale_feature_data(X_train, X_test)
```

```python
# train a decision tree classifier4
#model.load_decision_tree(max_depth=2)

model.load_knn(n_neighbors=5)
model.train(X_train_scaled, y_train)
#model.train(X_train, y_train)

#training results
train_results = model.predict(X_train_scaled, y_train)
print("Training Results:")
print(f"Accuracy: {train_results['accuracy']}")
print(f"Precision: {train_results['precision']}")
print(f"Recall: {train_results['recall']}")
print(f"F1 Score: {train_results['f1_score']}")
print(f"ROC AUC: {train_results['roc_auc']}")
print(f"Confusion Matrix:\n{train_results['confusion_matrix']}")
```

```python
#test results
test_results = model.predict(X_test_scaled, y_test)
#test_results = model.predict(X_test, y_test)
print("\nTest Results:")
print(f"Accuracy: {test_results['accuracy']}")
print(f"Precision: {test_results['precision']}")
print(f"Recall: {test_results['recall']}")
print(f"F1 Score: {test_results['f1_score']}")
print(f"ROC AUC: {test_results['roc_auc']}")
print(f"Confusion Matrix:\n{test_results['confusion_matrix']}")
```

```python
param_grid = {
    "n_neighbors": [1, 3, 5, 7, 9, 11, 13, 15],
    "weights": ["uniform", "distance"]
}

model.load_knn()

grid_results = model.grid_search(X_train_scaled, y_train, param_grid, cv=5, scoring="accuracy")

print("Best parameters:")
print(grid_results["best_params"])

print("\nBest CV accuracy:")
print(grid_results["best_score"])

# test the knn classifier
results = model.predict(X_test_scaled, y_test)
print("KNN Classifier Results:")
print(f"Accuracy: {results['accuracy']}")
print(f"Precision: {results['precision']}")
print(f"Recall: {results['recall']}")
print(f"F1 Score: {results['f1_score']}")
print(f"ROC AUC: {results['roc_auc']}")
print(f"Confusion Matrix:\n{results['confusion_matrix']}")
```

```python
# grid search for hyperparameter tuning
param_grid = {
    "colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "reg_alpha": [0, 0.01, 0.1, 1],
    "reg_lambda": [1, 2, 5, 10]
}

# load the xgboost model with specified hyperparameters found as best results
model.load_xgboost(learning_rate=0.1, n_estimators=150, max_depth=2, subsample=0.8, min_child_weight=7, gamma=0)

grid_results = model.grid_search(X_train, y_train, param_grid, cv=5, scoring="accuracy", n_jobs=-1)
print("Best parameters:")
print(grid_results["best_params"])

print("\nBest CV accuracy:")
print(grid_results["best_score"])

# test the xgboost classifier
results = model.predict(X_test, y_test)
print("Xgboost Classifier Results:")
print(f"Accuracy: {results['accuracy']}")
print(f"Precision: {results['precision']}")
print(f"Recall: {results['recall']}")
print(f"F1 Score: {results['f1_score']}")
print(f"ROC AUC: {results['roc_auc']}")
print(f"Confusion Matrix:\n{results['confusion_matrix']}")

```

```python
#DECISION TREE CLASSIFIER RESULTS

# grid search for hyperparameter tuning
param_grid = {
    'max_depth': [1, 2, 3, 4],
    'min_samples_split': [2, 5, 10, 15, 20],
    'min_samples_leaf': [1, 2, 3, 5, 10]
}

model.load_decision_tree()

grid_results = model.grid_search(X_train, y_train, param_grid, cv=5)
print("Best parameters:")
print(grid_results["best_params"])

print("\nBest CV accuracy:")
print(grid_results["best_score"])

# test the decision tree classifier
results = model.predict(X_test, y_test)
print("\nDecision Tree Classifier Results:")
print(f"Accuracy: {results['accuracy']}")
print(f"Precision: {results['precision']}")
print(f"Recall: {results['recall']}")
print(f"F1 Score: {results['f1_score']}")
print(f"ROC AUC: {results['roc_auc']}")
print(f"Confusion Matrix:\n{results['confusion_matrix']}")

"""The initial Decision Tree exhibited substantial overfitting, 
achieving 95.45% training accuracy but only 45.45% test accuracy. 
Hyperparameter tuning using 5-fold cross-validation identified a shallow tree (max_depth=2) 
with min_samples_leaf=10 as the best configuration within the tested parameter space. 
The tuned model achieved a mean cross-validation accuracy of 56.93% and a test accuracy of 59.09%. 
Its ROC-AUC of 0.669 indicates that the model provides some discriminatory ability between the two classes, 
although performance remains limited."""
```
