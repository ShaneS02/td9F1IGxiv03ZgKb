```python
from models.CoreModel import CoreModel
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
model = CoreModel()
model.load_data_from_csv("../data/ACME-HappinessSurvey2020.csv")
model.preprocess_data()

    # split the dataset into training and testing sets
X_train, X_test, y_train, y_test = model.split_data()
X_train_scaled, X_test_scaled = model.scale_feature_data(X_train, X_test)
```

```python
sns.countplot(data=model.get_data(), x="Y")
plt.title("Distribution of Customer Happiness")
plt.show()

#data is not imbalanced, so we can use accuracy as a metric to evaluate the model's performance
```

```python
def plot_feature_distribution(feature_name):
    sns.countplot(data=model.get_data(), x=feature_name)
    plt.title(f"Distribution of {feature_name}")
    plt.show()

for feature in model.get_data().columns:
    if feature != "Y":
        plot_feature_distribution(feature)

"""Summary of all the graphs

X1, X4 - X6: The distribution is left-skewed, 
with most values concentrated on the upper end of the scale. 
There are a few outliers on the lower end.

X2: The distribution of X2 is right-skewed,
with most values concentrated on the lower end of the scale.
There are a few outliers on the higher end.

X3: The distribution of X3 is approximately normal,
with most values concentrated around the mean and fewer values on the tails.


"""
```

```python
def plot_feature_boxplot(feature_name):
    sns.boxplot(y=model.get_data()[feature_name])
    plt.title(f"Boxplot of {feature_name}")
    plt.show()

for feature in model.get_data().columns:
    if feature != "Y":
        plot_feature_boxplot(feature)
```

```python
def plot_feature_boxplot_by_target(feature_name):
    sns.boxplot(data=model.get_data(), x="Y", y=feature_name)
    plt.title(f"Boxplot of {feature_name} by Target Variable Y")
    plt.show()

for feature in model.get_data().columns:
    if feature != "Y":
        plot_feature_boxplot_by_target(feature)
```
